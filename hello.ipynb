## Etape 3 : Affichage de variables

```java
// 1. Déclaration et initialisation des variables
String nomProduit1 = "Stylos";
double prixUnitaire1 = 1.20;
int quantite1 = 5;

String nomProduit2 = "Cahiers";
double prixUnitaire2 = 3.50;
int quantite2 = 2;

// Calcul du total général
double total = (prixUnitaire1 * quantite1) + (prixUnitaire2 * quantite2);

// 2. Affichage avec formatage (printf)
System.out.println("+-----------+--------+-----+--------+");
System.out.println("| Produit   | PU     | Qté | Prix   |");
System.out.println("+-----------+--------+-----+--------+");

// %-9s = texte aligné à gauche sur 9 espaces
// %.2f = nombre à virgule avec 2 chiffres après la virgule
// %d   = nombre entier
// %n   = saut de ligne

System.out.printf("| %-9s | %.2f € |  %d  | %.2f €|%n", nomProduit1, prixUnitaire1, quantite1, prixUnitaire1 * quantite1);
System.out.printf("| %-9s | %.2f € |  %d  | %.2f €|%n", nomProduit2, prixUnitaire2, quantite2, prixUnitaire2 * quantite2);

System.out.println("+-----------+--------+-----+--------+");
System.out.printf("| TOTAL     |        |     |%.2f €|%n", total);
System.out.println("+-----------+--------+-----+--------+");
```

## Etape 4: Saisie des informations

```java
// On initialise le scanner 
Scanner scanner = new Scanner(System.in);

// Déclaration des produits (prix fixes, mais quantités à saisir)
String nomProduit1 = "Stylos";
double prixUnitaire1 = 1.20;

String nomProduit2 = "Cahiers";
double prixUnitaire2 = 3.50;

// Demande à l'utilisateur
System.out.print("Combien de " + nomProduit1 + " voulez-vous ? ");
int quantite1 = scanner.nextInt(); // Le programme attend que tu tapes un nombre ici

System.out.print("Combien de " + nomProduit2 + " voulez-vous ? ");
int quantite2 = scanner.nextInt(); // Le programme attend le deuxième nombre

// Calcul du nouveau total
double total = (prixUnitaire1 * quantite1) + (prixUnitaire2 * quantite2);

// Affichage du tableau (Identique à l'étape 3)
System.out.println("\n+-----------+--------+-----+--------+");
System.out.println("| Produit   | PU     | Qté | Prix   |");
System.out.println("+-----------+--------+-----+--------+");

System.out.printf("| %-9s | %.2f € |  %d  | %.2f €|%n", nomProduit1, prixUnitaire1, quantite1, prixUnitaire1 * quantite1);
System.out.printf("| %-9s | %.2f € |  %d  | %.2f €|%n", nomProduit2, prixUnitaire2, quantite2, prixUnitaire2 * quantite2);

System.out.println("+-----------+--------+-----+--------+");
System.out.printf("| TOTAL     |        |     |%.2f €|%n", total);
System.out.println("+-----------+--------+-----+--------+");
```
